🍉 + 🍓 + 🥭 = 12

````python
# Setup variables
strawberry = Int('strawberry')
grape = Int('grape')
watermelon = Int('watermelon')
mango = Int('mango')    
result = Int('result')
 
    
# Setup an expression that could be fed to a solver
conditions = [
    strawberry + grape * strawberry == 55,
    watermelon + strawberry + mango == 12,
    mango + grape - watermelon == 9,
    grape - mango + watermelon == 11,
    mango + watermelon * grape == result
    
]

s = Solver()
s.add(conditions)
is_sat = s.check()  # check if solution exist


solution = s.model()
print("ANSWER IS:", solution[result])
print("Strawberry:", solution[strawberry])
print("Grape:", solution[grape])
print("Watermelon:", solution[watermelon])
print("Mango:", solution[mango])
````

**Примечание:**

- Нужно обязательно использовать все доступные камни

На позиции `(x=0, y=1)` поднимаемся 1 раз,

2. Дан набор камней, `n = 3`
```

  00 01 02 03 04 05 06 07 08 09 10 11 12
```

**Решение**   
```

2|            🪨
1|🪨 🪨    🪨 🪨
0|-- -- -- -- -- -- -- -- -- -- -- -- -- 
  00 01 02 03 04 05 06 07 08 09 10 11 12
```
Как мы видим, в данном случае мы можем выстроить камни так, чтобы путь включал в себя подъем только на `3` клетки и спуск только на `3` клетки.  
На позиции `(x=0, y=1)` поднимаемся 1 раз,     
На позиции `(x=1, y=1)` спускаемся 1 раз,   
На позиции `(x=3, y=1)` поднимаемся 1 раз,   
На позиции `(x=4, y=2)` поднимаемся 1 раз,   
На позиции `(x=4, y=2)` спускаемся 2 раза,   
Итого - `3` подъема и `3` спуска.


````python
from collections import Counter

def solve_stones(stones, n):
    if any(size not in (1, 2, 3) for size in stones):
        raise ValueError("Stone heights must be 1, 2, or 3")

    if not stones:
        return [] if n == 0 else None

    m = len(stones)
    width = 2 * m - 1

    solver = Solver()

    # Height of every column (0 means empty).
    h = [Int(f"h_{i}") for i in range(width)]

    # 1. Allowed heights.
    for cell in h:
        solver.add(And(cell >= 0, cell <= 3))

    # 2. Use every stone exactly once.
    counts = Counter(stones)

    for size in (1, 2, 3):
        solver.add(
            Sum([If(cell == size, 1, 0) for cell in h])
            == counts[size]
        )

    # Include the ground at both ends.
    path = [IntVal(0), *h, IntVal(0)]

    # 3. Total ascent.
    ascent = Sum([
        If(right > left, right - left, 0)
        for left, right in zip(path, path[1:])
    ])

    # 4. Total descent.
    descent = Sum([
        If(left > right, left - right, 0)
        for left, right in zip(path, path[1:])
    ])

    solver.add(ascent == n)
    solver.add(descent == n)

    result = solver.check()

    if result == sat:
        model = solver.model()
        return [model.eval(cell).as_long() for cell in h]

    if result == unsat:
        return None

    raise RuntimeError(f"Z3 returned {result}")


# Test cases
print(solve_stones([1, 2, 1, 1, 2, 2], 2))
print(solve_stones([1, 2, 1, 1], 3))
print(solve_stones([2, 2], 3))
````

# Задача №4 - Кабельные войны
В серверной существует n-ое количество серверов с разной информацией внутри (красная, синяя и бордовая) и с разными требованиями по информации которая должны через них проходить (красная, синяя или бордовая). Также есть ограниченное количество кабелей (красные, синие и бордовые).
Каждый сервер может проводить не ограниченное количество цветов (если не сказано иное), быть подключенным к одному кабелю каждого цвета, один кабель может соединять ровно два сервера, красная информация течет только по красному кабелю и т.д. Если, например, бордовый кабель подключен к серверу содержащему красный цвет, то поток бордовых данных всё равно ни откуда не возьмется. Необходимо узнать конфигурацию при которой можно было бы соединить все сервера с соблюдением требований по информации которая должны через них проходить, с учетом следующих особенностей кабелей.
- Если к серверу подключены синий и бордовый кабель он не может далее передавать красную информацию
- Если к серверу подключен только бордовый кабель он может передать бордовый сигнал только следующему серверу, а тот в свою очередь не может далее передавать бордовый сигнал (если только он не содержит бордовую информацию)

**Входные данные:**
* Сервер a - информация внутри (синий), требование по цвету (красный)
* Сервер b - информация внутри (красный), требование по цвету (бордовый)
* Сервер c - информация внутри (красный), требование по цвету (красный)
* Сервер d - информация внутри (бордовый), требование по цвету (бордовый)
* Сервер e - информация внутри (бордовый), требование по цвету (красный)
* Сервер f - информация внутри (синий), требование по цвету (бордоый)


* Красные кабеля -  3 шт
* Синие кабеля   -  4 шт
* Бордовые кабеля - 5 шт


````python

from itertools import combinations, permutations
from z3 import (
    And, Bool, If, Implies, Int,
    Not, Optimize, Or, Sum, is_true, sat, unsat
)

# R = red, B = blue, M = maroon
# server: (information stored, information required)

servers = {
    "a": ("B", "R"),
    "b": ("R", "M"),
    "c": ("R", "R"),
    "d": ("M", "M"),
    "e": ("M", "R"),
    "f": ("B", "M"),
}

limits = {"R": 2, "B": 1, "M": 2}

names = list(servers)
colors = tuple(limits)

opt = Optimize()

# 1. Physical cable connections (undirected)
E = {
    (u, v, color): Bool(f"E_{u}_{v}_{color}")
    for u, v in combinations(names, 2)
    for color in colors
}

def edge(u, v, color):
    return E[(*sorted((u, v)), color)]


# 2. Data transfers (directed)
T = {
    (u, v, color): Bool(f"T_{u}_{v}_{color}")
    for u, v in permutations(names, 2)
    for color in colors
}

def has_cable(u, color):
    return Or(*[
        edge(u, v, color)
        for v in names if v != u
    ])


# 3. At most one cable of each color per server
for u in names:
    for color in colors:
        opt.add(
            Sum(*[
                If(edge(u, v, color), 1, 0)
                for v in names if v != u
            ]) <= 1
        )


# 4. Limited cable supply
for color, limit in limits.items():
    opt.add(
        Sum(*[
            If(link, 1, 0)
            for (u, v, c), link in E.items()
            if c == color
        ]) <= limit
    )


# 5. Data cannot appear from nowhere.
# Under our assumptions, the sender must
# contain the corresponding color.
for (u, v, color), transfer in T.items():
    opt.add(
        Implies(transfer, edge(u, v, color))
    )

    if servers[u][0] != color:
        opt.add(Not(transfer))


# 6. Blue + maroon cables block outgoing red data
for u in names:
    block_red = And(
        has_cable(u, "B"),
        has_cable(u, "M")
    )

    for v in names:
        if u != v:
            opt.add(
                Implies(
                    block_red,
                    Not(T[u, v, "R"])
                )
            )


# 7. Maroon-only cable rule
for u in names:
    only_maroon = And(
        has_cable(u, "M"),
        Not(has_cable(u, "R")),
        Not(has_cable(u, "B"))
    )

    for v in names:
        if u == v or servers[v][0] == "M":
            continue

        for w in names:
            if w not in (u, v):
                opt.add(
                    Implies(
                        And(
                            only_maroon,
                            T[u, v, "M"]
                        ),
                        Not(T[v, w, "M"])
                    )
                )


# 8. Every server must receive its required data
for v in names:
    own, required = servers[v]

    if own != required:
        opt.add(
            Or(*[
                T[u, v, required]
                for u in names if u != v
            ])
        )


# 9. All servers must form one connected graph
rank = {
    u: Int(f"rank_{u}")
    for u in names
}

opt.add(rank[names[0]] == 0)

for v in names:
    opt.add(
        rank[v] >= 0,
        rank[v] < len(names)
    )

    if v != names[0]:
        opt.add(
            Or(*[
                And(
                    Or(*[
                        edge(u, v, c)
                        for c in colors
                    ]),
                    rank[u] < rank[v]
                )
                for u in names if u != v
            ])
        )


# 10. Minimize cables and data transfers
opt.minimize(
    Sum(*[
        If(link, 1, 0)
        for link in E.values()
    ])
)

opt.minimize(
    Sum(*[
        If(transfer, 1, 0)
        for transfer in T.values()
    ])
)


# 11. Find a solution
result = opt.check()

if result == sat:
    model = opt.model()

    print("SAT: valid configuration found")

    print("\nCables:")
    for (u, v, color), link in E.items():
        if is_true(model.eval(link)):
            print(f"  {u} --{color}-- {v}")

    print("\nData transfers:")
    for (u, v, color), transfer in T.items():
        if is_true(model.eval(transfer)):
            print(f"  {u} --{color}--> {v}")

elif result == unsat:
    print("UNSAT: no valid configuration")

else:
    print("UNKNOWN:", opt.reason_unknown())
````

## 2. Логика получения решения

Начнем со сложения чисел по разрядам.

Введем переменные переноса `c1`, `c2`, `c3`, `c4` (каждая равна 0 или 1).

Получим систему ограничений:

\\[ \begin{aligned} D+E &= Y+10c_1\\\ N+R+c_1 &= E+10c_2\\\ E+O+c_2 &= N+10c_3\\\ S+M+c_3 &= O+10c_4\\\ c_4 &= M \end{aligned} \\]

### Шаг 1. Определяем M и O

Сумма двух четырехзначных чисел дает пятизначное число. Следовательно, в старший разряд переносится единица:

\\[ \boxed{M=1} \\]

Для тысяч:

\\[ S+1+c_3=O+10 \\]

Поскольку `S ≤ 9`, `c3 ≤ 1`, получаем `O ≤ 1`. Но `O ≠ M`, следовательно:

\\[ \boxed{O=0} \\]

Тогда:

\\[ S+c_3=9 \\]

Возможны два случая: `S = 9, c3 = 0` или `S = 8, c3 = 1`.

### Шаг 2. Определяем S и N

Рассмотрим сотни:

\\[ E+0+c_2=N+10c_3 \\]

Если `c3 = 1`, то единственный возможный вариант `E = 9`, `c2 = 1`, `N = 0`. Но `O = 0`, а буквы должны соответствовать разным цифрам.

Значит, `c3 = 0`, откуда:

\\[ \boxed{S=9} \\]

Теперь равенство для сотен:

\\[ E+c_2=N \\]

Поскольку `E ≠ N`, перенос `c2` не может быть нулем. Поэтому:

\\[ \boxed{c_2=1,\quad N=E+1} \\]

### Шаг 3. Определяем R

Для десятков:

\\[ N+R+c_1=E+10 \\]

Подставим `N = E + 1`:

\\[ R+c_1=9 \\]

Цифра 9 уже занята буквой S. Поэтому `R = 9` невозможно, значит:

\\[ \boxed{R=8,\quad c_1=1} \\]

### Шаг 4. Определяем E, N, D, Y

Остались свободные цифры:

\\[ \\{2,3,4,5,6,7\\} \\]

Знаем:

\\[ N=E+1,\qquad D+E=Y+10 \\]

Поскольку `Y ≥ 2`, нужно `D + E ≥ 12`.

- Если `E ≤ 4`, то `D + E ≤ 11`, не подходит.
- Если `E = 6`, то `N = 7`, и максимальное доступное `D = 5`. Сумма снова меньше 12.
- Следовательно, `E = 5`, `N = 6`. Тогда единственная возможность: `D = 7`, `Y = 2`.

Таким образом, получаем окончательный ответ:

\\[ \boxed{9567+1085=10652} \\]

#### Решение задачи о восьми ферзях

Задача: разместить 8 ферзей на шахматной доске 8×8 так, чтобы никакие два ферзя не атаковали друг друга.

Ферзь атакует любую клетку на одной с ним горизонтали, вертикали или диагонали.

##### 1. Конкретное решение

Ферзи стоят на клетках:

a1, b5, c8, d6, e3, f7, g2, h4.

###### 2. Логика поиска решения

Представим расстановку как массив из восьми чисел:

```
queens = [1, 5, 8, 6, 3, 7, 2, 4]
```

Индекс элемента обозначает столбец, а его значение — строку, в которой находится ферзь.

Для поиска можно использовать перебор с возвратом (backtracking):

1. В каждом столбце размещаем ровно одного ферзя, исключая атаки по вертикали.
2. Выбираем строку, которая ещё не занята, исключая атаки по горизонтали.
3. Проверяем, не атакует ли новая фигура уже поставленных ферзей по диагонали.
4. Если ни одна строка не подходит, возвращаемся к предыдущему столбцу и пробуем другой вариант.
5. Продолжаем, пока не разместим все восемь фигур.

Такая процедура позволяет найти указанную расстановку.

###### 3. Проверка результата

Два ферзя на клетках \\((x_1,y_1)\\) и \\((x_2,y_2)\\) не должны удовлетворять ни одному из условий:

\\[ x_1=x_2 \\]

\\[ y_1=y_2 \\]

\\[ |x_1-x_2|=|y_1-y_2| \\]

Последнее условие описывает атаку по диагонали.

Для выбранной расстановки:

- Вертикали: все ферзи стоят в разных столбцах.
- Горизонтали: значения `[1, 5, 8, 6, 3, 7, 2, 4]` не повторяются.
- Диагонали: у всех ферзей различаются как значения \\(y-x\\), так и значения \\(y+x\\).

Проверим диагонали численно:

| Ферзь | \\(y-x\\) | \\(y+x\\) |
| ----- | --------- | --------- |
| a1    | 0         | 2         |
| b5    | 3         | 7         |
| c8    | 5         | 11        |
| d6    | 2         | 10        |
| e3    | −2        | 8         |
| f7    | 1         | 13        |
| g2    | −5        | 9         |
| h4    | −4        | 12        |

Ни в одном из двух последних столбцов нет повторяющихся значений. Следовательно, ни одна пара ферзей не находится на одной диагонали.

Решение найдено:

\\[ \boxed{[1,\ 5,\ 8,\ 6,\ 3,\ 7,\ 2,\ 4]} \\]

### 2. Логика построения решения

Вместо случайного перебора построим таблицу по определённому правилу.

Шаг 1. Заполняем первую строку цифрами от 1 до 9:

`1 2 3 4 5 6 7 8 9`

Шаг 2. Для второй строки циклически сдвигаем последовательность влево на 3 позиции:

`4 5 6 7 8 9 1 2 3`

Шаг 3. Для третьей строки повторяем сдвиг на 3 позиции:

`7 8 9 1 2 3 4 5 6`

Так мы получаем первые три строки, в которых каждый квадрат 3×3 содержит все цифры от 1 до 9.

Шаг 4. При переходе к четвёртой строке делаем сдвиг относительно третьей строки на 3 позиции вправо (что эквивалентно сдвигу первой строки на 1 позицию влево):

`2 3 4 5 6 7 8 9 1`

Шаг 5. Повторяем эту схему: внутри каждой группы из трёх строк сдвигаем цифры на 3 позиции, а в начале следующей группы смещаем начальную последовательность ещё на одну позицию.

В результате получаем всю таблицу.

### 3. Математическая формула решения

Если пронумеровать строки и столбцы от 0 до 8, то значение каждой клетки можно вычислить:

\\[ \boxed{a\_{r,c}=1+\big((3(r\bmod3)+\lfloor r/3\rfloor+c)\bmod9\big)} \\]

Где:

- \\(r\\) — номер строки;
- \\(c\\) — номер столбца;
- \\(\bmod\\) — остаток от деления;
- \\(\lfloor r/3\rfloor\\) — целая часть от деления.

Эта формула позволяет получить всю таблицу без поиска и перебора.

## Решение задачи о 100 животных

Ответ:

Итого: 100 животных

## $100

### 1. Составим систему уравнений

Обозначим:

- \\(d\\) — количество собак;
- \\(c\\) — количество котов;
- \\(m\\) — количество мышей.

По условию необходимо купить ровно 100 животных:

\\[ d+c+m=100 \\]

И потратить ровно 100 долларов:

\\[ 15d+c+0.25m=100 \\]

Причём \\(d,c,m\\) — положительные целые числа.

### 2. Упростим систему

Умножим второе уравнение на 4, чтобы избавиться от дробей:

\\[ 60d+4c+m=400 \\]

Вычтем из него первое уравнение:

\\[ (60d+4c+m)-(d+c+m)=400-100 \\]

Получаем:

\\[ \boxed{59d+3c=300} \\]

### 3. Найдём количество собак

Поскольку \\(3c\\) и \\(300\\) делятся на 3, число \\(59d\\) тоже должно делиться на 3.

Так как 59 не делится на 3, количество собак должно быть кратно 3.

При этом больше пяти собак купить невозможно, потому что:

\\[ 6\cdot15=90 \\]

После покупки шести собак останется только $10, но нужно купить ещё 94 животных, что невозможно даже по минимальной цене $0,25 за животное.

Следовательно, единственное положительное число собак, кратное 3 и не превышающее 5:

\\[ \boxed{d=3} \\]

### 4. Найдём количество котов и мышей

Подставляем \\(d=3\\):

\\[ 59\cdot3+3c=300 \\]

\\[ 3c=123 \\]

\\[ \boxed{c=41} \\]

Оставшиеся животные — мыши:

\\[ m=100-3-41 \\]

\\[ \boxed{m=56} \\]

### 5. Проверка

Количество животных:

\\[ 3+41+56=\boxed{100} \\]

Стоимость:

\\[ 3\cdot15+41\cdot1+56\cdot0.25 \\]

\\[ 45+41+14=\boxed{100} \\]

Каждого вида куплено хотя бы одно животное.

Итог: необходимо купить 3 собаки, 41 кота и 56 мышей.

Это единственное решение в положительных целых числах. Мы не просто подобрали числа, а доказали их единственность с помощью системы линейных уравнений и свойства делимости на 3.
